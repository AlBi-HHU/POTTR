```python
import os
import altair as alt
import pandas as pd
```

```python
out_path = 'Output/benchmarking results/'
os.makedirs(out_path, exist_ok=True)
```

```python
path = 'Output/benchmarking/new_benchmarking_k_runtime/k_average_runtimes_and_peak_mem_traj_sizes.csv'
df = pd.read_csv(path)
df
```

```python
#[df['trajectory size'] > 1]
wall = alt.Chart(df).mark_line().encode(
    x=alt.X('k', axis=alt.Axis(labelFontSize=16, titleFontSize=16, title='Parameter k' )),
    y=alt.Y('wall-clock time', axis=alt.Axis(labelFontSize=16, titleFontSize=16, title='Wall-clock time (sec)')),
    color=alt.Color(
        'Graph count:O',
        scale=alt.Scale(
            domain=[128, 256],
            range=['#cb2a7b', '#98b2f3']
        ),
        title='Graph count',
        legend=alt.Legend(
            titleFontSize=16,
            labelFontSize=14,
            symbolSize=100
        )
    )
).properties(
    title=alt.TitleParams(
        text='Effect of parameter k on the runtime',
        fontSize=16
    ),
    width=280
)

#[df['trajectory size'] > 1]
rss = alt.Chart(df).mark_line().encode(
    x=alt.X('k', axis=alt.Axis(labelFontSize=16, titleFontSize=16, title='Parameter k' )),
    y=alt.Y('maximum resident set size', axis=alt.Axis(labelFontSize=16,
                                                       titleFontSize=16,
                                                       title='Maximum RSS (GB)')),
    color=alt.Color(
        'Graph count:O',
        scale=alt.Scale(
            domain=[128, 256],
            range=['#cb2a7b', '#98b2f3']
        ),
        title='Graph count',
        legend=alt.Legend(
            titleFontSize=16,
            labelFontSize=14,
            symbolSize=100
        )
    )
).properties(
    title=alt.TitleParams(
        text='Effect of parameter k on the RSS',
        fontSize=16
    ),
    width=280
)


rss.show()
rss.save(out_path + 'new_effect_of_k_rss.svg', 'svg')
```

```python
wall.show()
wall.save(out_path + 'new_effect_of_k_wall.svg', 'svg')
```

```python
traj_size = alt.Chart(df).mark_line().encode(
    x=alt.X('k', axis=alt.Axis(labelFontSize=16, titleFontSize=16, title='Parameter k' )),
    y=alt.Y('trajectory size', axis=alt.Axis(labelFontSize=16, titleFontSize=16, title='Trajectory size')),
    color=alt.Color(
        'Graph count:O',
        scale=alt.Scale(
            domain=[128, 256],
            range=['#cb2a7b', '#98b2f3']
        ),
        title='Graph count',
        legend=alt.Legend(
            titleFontSize=16,
            labelFontSize=14,
            symbolSize=100
        )
    )
).properties(
    title=alt.TitleParams(
        text='Output size',
        fontSize=16
    ),
    width=180
)


traj_size.show()
traj_size.save(out_path + 'new_traj_sizes_k_benchmark.svg', 'svg')
```
