```python
import os
import pandas as pd
import altair as alt
```

```python
df_list = []
path = '/Code/toy_example/toy_graphs_resolved'
for subdir in os.listdir(path):
    if subdir == '.DS_Store':
        continue
    instance = subdir.split('_')[0]
    for subsubdir in os.listdir(os.path.join(path, subdir, 'output')):
        print(subsubdir)
        if not os.path.isdir(os.path.join(path, subdir, 'output', subsubdir)):
            continue
        if subsubdir == '.DS_Store' or 'resample' in subsubdir or 'pottr' in subsubdir:
            continue
        df_ = pd.read_csv(os.path.join(path, subdir, 'output', subsubdir, 'significance_output.txt'), delimiter=';', index_col=False)
        df_['resolved_edges'] = int(instance)
        df_['true_edges'] = int(subsubdir)
        df_list.append(df_)
```

```python
df = pd.concat(df_list)
df
```

```python
x_axis = alt.X('true_edges', title=['Scenario', '(Graphs showing the resolved order as edges (of 10))'])
legend = alt.Color('resolved_edges:N', title='Instance')

chart1 = alt.Chart(df).mark_line().encode(
    x=x_axis,
    y=alt.Y('support_score', title=['Frequency of trajectory edges present in data']),
    color=legend,
)

chart2 = alt.Chart(df).mark_line().encode(
    x=x_axis,
    y=alt.Y('resolution_evidence', title=['Resolution score', '(random cluster order fits trajectory)']),
    color=legend,
)

chart = alt.concat(chart1, chart2)
chart.show()
chart.save('resolution_score.svg', 'svg')
```
