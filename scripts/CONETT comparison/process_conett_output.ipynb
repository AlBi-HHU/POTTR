```python
import re
import read_input_dags
import networkx as nx
from IPython.display import Image, display
```

```python
file_lung = '/CONETT-master/output/sara_test_trees_lung_te2_a1_t2_e2.000000/subnetworks/1_tree_adj.txt'
file_aml = '/CONETT-master/output/sara_test_trees_aml_te2_a1_t2_e2.000000/subnetworks/1_tree_adj.txt'
```

```python
def process_file(file_path):
    with open(file_path, 'r') as f:
        content = f.read()

    segments = content.strip().split('\n\n')
    last_lines = segments[-1].strip().split('\n')
    return last_lines
```

```python
def create_graph(last_lines):
    G = nx.DiGraph()
    for line in last_lines:
        line = line.strip()
        line = re.sub(r' \(\d+\)', '', line)
        parts = line.split(' ')
        for i, part in enumerate(parts):
            part = part.strip()
            if i % 2:
                w = int(re.match(r'--\[(\d+)\]-->', part).group(1))
                if w > 1:
                    G.add_edge(parts[i-1], parts[i+1], weight=w)
    dot = nx.drawing.nx_pydot.to_pydot(G)
    # created with chatGPT
    # Force nice layout
    dot.set_rankdir("LR")  # left → right

    for edge in dot.get_edges():
        u = edge.get_source().strip('"')
        v = edge.get_destination().strip('"')
        if G.has_edge(u, v):
            w = G[u][v].get("weight")
            if w is not None:
                edge.set_label(str(w))

    # Show in notebook
    png = dot.create_png()
    display(Image(png))
    return G
```

```python
lines = process_file(file_aml)
create_graph(lines)
```
