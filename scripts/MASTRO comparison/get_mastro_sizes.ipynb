```python
def extract_nodes(edge_string):
    nodes = set()
    edges = edge_string.strip('[]').split(',')
    for e in edges:
        if '->-' in e:
            n1, n2 = e.split('->-')
            nodes.add(n1)
            nodes.add(n2)
        elif '-?-' in e:
            n1, n2 = e.split('-?-')
            nodes.add(n1)
            nodes.add(n2)
        elif '-/-' in e:
            n1, n2 = e.split('-/-')
            nodes.add(n1)
            nodes.add(n2)
        else:
            print('Unknown edge type', e)

    nodes.remove('g')
    return nodes
```

```python
sizes_dict = dict()
for edge_list, occ_list in zip(df_mastro_full['edges_traj'], df_mastro_full['traj_occ_list']):
    nodes = extract_nodes(edge_list)
    size = len(nodes)
    occ = set(occ_list.strip('[]').split(', '))
    k = len(occ)
    if k not in sizes_dict:
        print(k, size, occ, nodes)
        sizes_dict[k] = size
    else:
        if size > sizes_dict[k]:
            print(k, size, occ, nodes)
            sizes_dict[k] = size

sizes_dict
```
